```python
#utilizando dicionário no pyhton
# o diocionário né diferente, pois para acessar os valores é necessáirio utilizar a chave, ele é formado por par de informações chave: valor
# não é possível utilizar mais de um valor para a mesma chave, caso isso aconteçã ele ira ignorar um dos valores e considerar somente um

mais_vendidos = { 'tecnologia':'iphone', 'refrigeracao':'ar consul 12000btu', 'livros':'o alquimista'}
vendas_tecnologia = {'iphnoe':15000, 'samsung galxy':12000, 'ps5':14300, 'tablet':1720}

#acessando o valor com a chave do dicionário
qtd_iphone = vendas_tecnologia['ps5']
print(qtd_iphone)
print(f'O livro mais vendido foi {mais_vendidos['livros']}')
print(f'O produto mais vendido em refrigeração foi {mais_vendidos['refrigeracao']}')

#utilizando o metodo get ( a difernça é que o método get retorna o valor {None} caso não encontre a chave indicada no dicionário)
print(f'O livro mais vendido foi {mais_vendidos.get('livros')}')
print(f'O produto mais vendido em refrigeração foi {mais_vendidos.get('refrigeracao')}')


#verificando se um item está dentro do dicinário com o if
if 'tablet' in vendas_tecnologia:
  print(vendas_tecnologia['tablet'])
else:
  print('O produto não está dentro dos listado')

#verificando a existem de um item dentro do dicionário com o método get
if vendas_tecnologia.get('tablet') == None:
  print('O produto não está dentro da lista')
else:
  print(f'O preço do produto é igual a {vendas_tecnologia.get('tablet')}')
```
